```python
import pandas as pd

df = pd.read_csv('content_refresh_anonymized.csv')

#help suggest which pages need reviewing

print(f"impressions{df["impression_tier"].unique()},trends{df["trend_direction"].unique()}")
```

Suggest how many pages have down trending impressions.

54% of all pages

```python
df["down_trend"] = df["trend_direction"] == "down"
df["down_trend"].mean()
```

```python
df["low_impressions"] = df["impression_tier"] == "low"
df["low_impressions"].mean()
```

Of pages without low impressions 59% are trending down.
Of pages with low impressions 45% are trending down.
This is where other factors come into play because it is observed that not all low tier impressions have a down trend which suggests there are other factors that have a say in this outcome. This alone suggests that we can use those other factors to determine what actions can be taken.

```python
df.groupby("low_impressions")["down_trend"].mean()
```

Overall from the cells below you will observe pages with less users in the last 90 days have more down trending impressions .This suggest these pages need reviewing. My goal in this lane is to experiment on what factors are observed to these numbers and how we can use those factors based on the data to suggest decisions that can have a positive impact. For example a suggestion would be to increase the number of users. How can we do that? Do we need more articles? Do we need more word count? These are some of the observed actions that can be taken

```python
df.groupby(["low_impressions","down_trend"])["users_90d"].mean()
```
